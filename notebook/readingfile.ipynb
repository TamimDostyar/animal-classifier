# Importing libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import tensorflow as tf
```

```python
images_folder = os.path.join("data", "animals", "animals")
# List all animal categories
animal_categories = os.listdir(images_folder)
# Access a specific animal folder
first_animal = animal_categories[0] 
first_animal_path = os.path.join(images_folder, first_animal)
first_animal_images = os.listdir(first_animal_path)
# To iterate through all animals and their images
animal_stats = {}
for animal in animal_categories:
    animal_path = os.path.join(images_folder, animal)
    images = os.listdir(animal_path)
    animal_stats[animal] = len(images)

# Print animals with most and least images
sorted_stats = sorted(animal_stats.items(), key=lambda x: x[1])
```

```python
# convert txt to csv
animal_names = os.path.join("data", "animals.txt")
animal_names = pd.read_csv(animal_names)
# save as csv
animal_names.to_csv("data/animal_names.csv", index=False)

animal_names.head(20)
```

```python
# Creating a dataframe for animal images and seeing if they are in the csv
# Convert animal categories to DataFrame for easier comparison
df_animal_categories = pd.DataFrame(animal_categories, columns=["Animal Name"])

# Create empty dataframe for results
my_df = pd.DataFrame(columns=["Images Folder", "Animal Name"])

# Add each animal to the dataframe 
for animal in animal_categories:
    new_row = pd.DataFrame({"Images Folder": [animal], "Animal Name": [animal]})
    my_df = pd.concat([my_df, new_row], ignore_index=True)

my_df.head(20)
```
