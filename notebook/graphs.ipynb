Load process the data

```python
# Files locations
animal_names = pd.read_csv(os.path.join("data", "animal_names.csv"))
image_folders = os.listdir(os.path.join("data", "animals", "animals"))
number_images = len(image_folders)
files_locations = os.path.join("data", "animals/")

# Create a list of full paths by joining the base path with each image folder
folder_paths = [os.path.join(files_locations, image) for image in image_folders]

# Create the DataFrame with animal names and corresponding folder paths
image_ls = [image for image in image_folders]
animal_names_ls = [animal for animal in animal_names.iloc[:, 0].values]

#ensure all animals list is in image ls
# Create a DataFrame to store matching animals and their folders
matching_animals = []
for animal in sorted(animal_names_ls):
    for image in sorted(image_ls):
        if animal == image:
            matching_animals.append({"animal": animal, "folder": os.path.join(files_locations, image)})

# Create DataFrame from matching animals if any were found
if matching_animals:
    animal_dt_frame = pd.DataFrame(matching_animals)
    display(animal_dt_frame.head(90))
```
